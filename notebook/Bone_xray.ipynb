## 1. Import 

```python
import pandas as pd

from sklearn.model_selection import train_test_split
```

## 2. Loading data

```python
df = pd.read_excel('../data/dataset.xlsx')

df_drop = df
df_drop['gender'] =  df_drop['gender'].map({'M': 1, 'F': 0})
df_drop.drop(columns=['center'], inplace=True)
df_drop
```

```python
train_idx, test_idx = train_test_split(df_drop.index, test_size=0.25, random_state=42)

types = ['train' if i in train_idx else 'test' for i in df.index]

df_drop.insert(1, 'type', types)
display(df_drop.head())
```
